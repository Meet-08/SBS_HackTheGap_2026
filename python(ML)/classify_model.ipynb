```python
# Load your dataset
df = pd.read_csv("data/enriched_crop_data_v1.csv")


# --- STEP 1: ROBUST OUTLIER CLIPPING (P99) ---
# We cap values at the 99th percentile for each crop to handle data entry errors
# (e.g., Sannhamp 10,000 Q/Ha -> ~51 Q/Ha)
def clip_outliers(df, group_col="Crop", value_col="Yield_QHa"):
    # Calculate the P99 limit for each crop
    p99_limits = df.groupby(group_col)[value_col].quantile(0.99)

    # Map limits to the dataframe
    df["Yield_Limit"] = df[group_col].map(p99_limits)

    # Create a new clipped column (min of Value vs Limit)
    df["Yield_QHa_Clipped"] = df[[value_col, "Yield_Limit"]].min(axis=1)

    # Drop the helper column
    df.drop(columns=["Yield_Limit"], inplace=True)
    return df


df_clean = clip_outliers(df.copy())
print("Outlier clipping complete.")

# --- STEP 2: HIERARCHICAL MISSING VALUE IMPUTATION ---
# Strategy: District Median -> State Median -> Global Median
impute_cols = [
    "soil_ph",
    "soil_oc",
    "clay_pct",
    "sand_pct",
    "cec_cmol",
    "avg_temp",
    "humidity_avg",
    "rain_total",
    "solar_avg",
]

for col in impute_cols:
    # 1. Fill with Median of that District+Year (Most specific)
    df_clean[col] = df_clean[col].fillna(
        df_clean.groupby(["State", "District", "Year"])[col].transform("median")
    )

    # 2. Fill remaining with Median of that State+Year (Regional backup)
    df_clean[col] = df_clean[col].fillna(
        df_clean.groupby(["State", "Year"])[col].transform("median")
    )

    # 3. Fill remaining with Global Column Median (Final fallback)
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("Missing value imputation complete.")

# --- STEP 3: 6-MODEL SEGMENTATION ---
# Define the 6 distinct buckets to separate variance
model_mapping = {
    "Model_1_UltraLow": [
        "Jack Fruit",
        "Cardamom",
        "Niger seed",
        "Moth",
        "Sesamum",
        "Cowpea(Lobia)",
        "Horse-gram",
        "Linseed",
        "Sannhamp",
        "Black pepper",
        "Khesari",
    ],
    "Model_2_Low": [
        "Arhar/Tur",
        "Urad",
        "Gram",
        "Masoor",
        "Moong(Green Gram)",
        "Other Kharif pulses",
        "Other Rabi pulses",
        "Safflower",
        "Rapeseed &Mustard",
        "Jowar",
        "Small millets",
        "Ragi",
    ],
    "Model_3_Medium": [
        "Rice",
        "Wheat",
        "Maize",
        "Barley",
        "Bajra",
        "Castor seed",
        "Sunflower",
        "Groundnut",
        "Soyabean",
        "Coriander",
        "Dry chillies",
        "Peas & beans (Pulses)",
        "Cotton(lint)",
        "Other Cereals",
        "Tobacco",
    ],
    "Model_4_HighVeg": [
        "Potato",
        "Onion",
        "Garlic",
        "Ginger",
        "Dry Ginger",
        "Sweet potato",
        "Turmeric",
    ],
    "Model_5_UltraHigh": [
        "Banana",
        "Papaya",
        "Grapes",
        "Mesta",
        "Tapioca",
        "Sugarcane",
    ],
    "Model_6_Coconut": ["Coconut"],
}

# Invert dictionary for mapping
crop_to_model = {
    crop: model for model, crops in model_mapping.items() for crop in crops
}

# Apply Mapping
df_clean["Model_ID"] = df_clean["Crop"].map(crop_to_model)

# Handle Unclassified Crops (Assign to closest bucket based on their mean yield)
# For this dataset, we can map common remaining ones:
fallback_map = {
    "Jute": "Model_4_HighVeg",  # ~100 Q/Ha
    "Cashewnut": "Model_2_Low",  # ~10 Q/Ha
    "Arecanut": "Model_3_Medium",  # ~36 Q/Ha
    "Guar seed": "Model_2_Low",  # ~7 Q/Ha
    "Tomato": "Model_5_UltraHigh",  # ~160 Q/Ha (Borderline, but high variance)
    "Brinjal": "Model_4_HighVeg",  # ~118 Q/Ha
}
df_clean["Model_ID"] = df_clean["Model_ID"].fillna(df_clean["Crop"].map(fallback_map))

# Drop any remaining rows that couldn't be classified (likely very rare crops)
df_final = df_clean.dropna(subset=["Model_ID"])

# Save the processed file
df_final.to_csv("data/processed_crop_data.csv", index=False)
print("Data processing complete. Saved to 'processed_crop_data.csv'.")
print(df_final["Model_ID"].value_counts())
```
